List - Ordered & Changable(Mutable)

```python
fruits = ["Apple","banana","mango"]
print(fruits)
```

```python
print(fruits[0])
print(fruits[1])
```

```python
#change a value
fruits[2] = "orange"
print(fruits)
```

```python
a = [1,2,3.4,False,"word"]
print(a)
```

```python
#add an item to a list
a.append("apple")
print(a)
```

```python
#remove
a.remove("word")
print(a)
```

```python
#find no of elements
print(len(a))
```

```python
#for in list
fruits = ["Apple","banana","mango"]
for fruit in fruits:
    print(fruit)
```

**Tuple - Ordered & Unchangable(immutable)

```python
colors = ("red", "green", "blue")
print(colors)
print(type(colors))
```

```python
print(colors[0])
print(colors[1])
```

```python
#immutability
colors[1] = "yellow"
```

```python
for i in colors:
    print(i)
```

```python
colors.add("pink")
print(tuple)
```

```python
a = [1,2,3,4,5]
b = tuple(a)
print(type(a))
print(type(b))
print(a)
print(b)

```

Set - Unordered & unique

```python
students = {"Alice", "jhon", "sid"}
print(students)
print(type(students))
```

```python
students.add("david")
print(students)
```

```python
students.remove("Alice")
print(students)
```

```python
for student in students:
    print(student)
```

```python
list_1 = [1,1,2,3,3,5]
print(list_1)
```

```python
set_1 = {1,1,2,3,3,5}
print(set_1)
```

```python
list_2 = [10,9,9,8,5,1,1,2,4,8]
print(list_2)
list_2 = list(set(list_2))
print(list_2)
```

Dictionary- Key-Value data

```python
#dict with person details
person = {
    "name":"Alice",
    "Age": 25,
    "is_student": True
}
```

```python
person["Age"]
```

```python
print(type(person))
```

```python
person["city"] = "london"
print(person)
```

```python
person["Age"] = 50
print(person)
```

```python
for key, value in person.items():
    print(key, "->", value)
```
