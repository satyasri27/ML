2.Comaparison operators - Used to compare two values


```python
x = 5
y = 10
print (x == y)
print(x != y)
print(x >= y)
print(x <= y)
print(x > y)
print(x < y)
```

3.Logical Operators - Used to combine multiple conditions

```python
x = 10
y = 5
print(x!=y and x<y)
print(x!=y or x<y)
is_adult = True
has_id = False
print(is_adult and has_id)
print(is_adult or has_id)
print(not is_adult)
```

4.Assignment Operators - Used to assign values to variables
- = : Assign
- += : Add and Assign
- -= : Sub and Assign
- *= : Mul and Assign
- /= : Divide and Assign
- %= : Modulus and assign
- **=: square an assign
- //= : floor divide and assign


```python
number = 10
print(number)
number = number+100
print(number)
```

5.Membership Operators - Test for membership in a sequence:
- in : True if value is found
- not in : True if value is not found


```python
a = [1,2,3,4,5]
print( 3 in a)
print(10 in a)
print(3 not in a)
```

- is : True if both refer to same object
- is not : True if they do not refer to the same object


```python
a = c = [1,2,3]
b = [1,2,3]

print(a is b)
print(a is c)
print(a is not b)
```
