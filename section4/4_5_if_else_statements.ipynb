1. Basic if


```python
age = int(input("enter age"))
if age>=18 :
    print("eligible")
```

2.If-Else

```python
age = int(input("enter age"))
if age>=18:
    print("eligible")
else:
    print("not eligible")
```

3.If-ElseIf


```python
marks = int(input("enter marks"))
if marks >= 90:
    print("A")
elif  marks >= 80:
    print("B")
elif marks >=70:
    print("C")
```

5. If-Elif-Else

```python
marks = int(input("enter marks"))
if marks >= 90:
    print("A")
elif  marks >= 80:
    print("B")
elif marks >=70:
    print("C")
else:
    print("fail")
```

6.Nested-if

```python
age = int(input("enter age"))
has_id = input("do you have id? (yes/no): ")
if age>=18:
    if has_id == "yes":
        print("eligible")
    else:
        print("you need to show id")
else:
    print("not eligible")
```

7.Combining conditions with and/or

```python
age = int(input("enter age"))
has_id = input("do you have id? (yes/no): ")
if age>=18 and has_id == "yes":
     print("eligible")
else:
     print("not eligible")
```
