2.While Loop

```python
count = 3
while count <= 5:
    print("count is",count)
    count += 1
```

3.Loop Controls - break;continue

```python
for i in range(0,10):
    if i == 3:
        break
    print(i)
```

```python
for i in range(0,10):
    if i == 3:
        continue
    print(i)
```
