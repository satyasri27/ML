4.Use Class Attributes

Attributes are variables that belong to a class.Lets add and use attributes in dog class


```python
class Dog:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def bark(self):
        print("woof!")
```

```python
my_dog = Dog(name = "buddy", age =3)
print(my_dog.name)
print(my_dog.age)
my_dog.bark()
```

```python
dog_2 = Dog(name = "leo", age =4)
print(dog_2.name)
print(dog_2.age)
dog_2.bark()
```

5.Demonstrate Inheritance

Inheritance allows one class to use the properties and methods of another class

```python
class Dog:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def bark(self):
        print("woof!")
```

```python
#create a subclass that inherits from Dog
class Puppy(Dog):
    def play(self):
        print(self.name, "is playful")
```

```python
my_puppy = Puppy(name = "Max", age=1)
print(my_puppy.name)
print(my_puppy.age)
my_puppy.bark()
my_puppy.play()
```
