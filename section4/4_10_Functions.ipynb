2.Function with parameters

```python
def greet_user(name):
    print("hello"+name+"!")
greet_user(name = "sid")
```

```python
def add(a,b):
    result = a+b
    print("sum is: ", result)
add(a=5, b=10)
add(2,4)
```

3.Return a Value

```python
#using return
def square(n):
    return n*n
#save result in "result" variable
result = square(3)
print("square is: ", result)

#print o/p directly
print("square of 10 is :", square(10))
```

```python
#default parameters
def add(a=100, b=200):
    return a+b
add()
```
