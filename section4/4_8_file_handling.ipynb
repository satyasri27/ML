1.Opening and Reading a file

```python
#open a file in read mode
file = open("notes.txt","r")

#read the content
content = file.read()
print(content)
file.close()
```

2.Using with to read(auto-closing)

```python
with open("notes.txt","r") as file:
    content = file.read()
    print(content)
```

3.Writing to a file

```python
with open("notes.txt","w") as file:
    file.write("this is second file")
    
```
