DOT Product

```python
v1 = np.array([1,3])
v2 = np.array([4,2])
print("v1:", v1)
print("v2:", v2)
```

```python
dot = np.dot(v1, v2)
print(("Dot product v1.v2:", dot))
```

```python
print(dot)
```

Euclidean distance

```python
diff = v1-v2
print("difference v1-v2:",diff)
```

```python
distance = np.linalg.norm(diff) #use linalg.norm instead writing whole distancce formula

print("euclidean distance b/w v1 and v2:", distance)
```

Matrix multiplication and Linear Model

```python
X = np.array(
    [
        [1.0, 30.0],
        [2.0, 50.0],
        [3.0, 70.0]
    ]
)
print("X:\n", X)
print("X shape:", X.shape)
```

```python
w = np.array([
    [0.5],
    [0.1]
])
print("\nw:\n", w)
print("w shape:", w.shape)
```

```python
Xw = X @ w #dot
print("Xw:")
print(Xw)
print("Xw.shape:",Xw.shape)
```

y=Xw+b

```python
b=2.0
print("\nBias b:", b)
```

```python
y = Xw + b
print("\nPredictions y = Xw + b:\n",y)
print("y shape:",y.shape)
```

Connecting this to a sample dataset

```python
data = np.array([
    [86, 26.6, 31, 0],
    [183, 23.3, 32, 1],
    [89, 28.1, 21, 0],
    [137, 43.1, 33, 1],
])
print("full data matrix:\n", data)
print("data shape:", data.shape)
```

```python
x_features = data[:, :-1] #all cols except last col
y_target = data[:, -1] #only last col
print("\nFeature matrix X:\n", x_features)
print("X shape: ", x_features.shape)

print("\nTarget vector y:\n", y_target)
print("y shape: ", y_target.shape)
```
