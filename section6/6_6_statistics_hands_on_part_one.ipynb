Distribution curve

```python
col = df["total_bill"]

mean_val = np.mean(col)
median_val = np.median(col)
mode_val = col.mode()[0]
print(f"mean of total values: {mean_val:.2f}")
print(f"median of total values: {median_val:.2f}")
print(f"mode of total values: {mode_val:.2f}")

#visualise the histogram to understand the distribution of "total_bill" column
plt.figure(figsize=(5,3))
sns.histplot(col, kde=True, color="skyblue")
plt.axvline(mean_val, color="red", linestyle="--", label=f"Mean: {mean_val}")
plt.axvline(median_val, color="green", linestyle="-", label=f"Medeian: {median_val}")
plt.legend()
plt.title("Distribution of total bill")
plt.show()
```

```python
col = df["tip"]

mean_val = np.mean(col)
median_val = np.median(col)
mode_val = col.mode()[0]
print(f"mean of total values: {mean_val:.2f}")
print(f"median of total values: {median_val:.2f}")
print(f"mode of total values: {mode_val:.2f}")

#visualise the histogram to understand the distribution of "total_bill" column
plt.figure(figsize=(5,3))
sns.histplot(col, kde=True, color="skyblue")
plt.axvline(mean_val, color="red", linestyle="--", label=f"Mean: {mean_val}")
plt.axvline(median_val, color="green", linestyle="-", label=f"Medeian: {median_val}")
plt.legend()
plt.title("Distribution of tip with central tendency")
plt.show()
```

Measure of Spread

```python
#range
data_range = col.max() - col.min()

#variance
variance = np.var(col)
#std_dev
std_dev = np.std(col)
#IQR- Inter Quartile Range
Q1 = np.percentile(col, 25)
Q3 = np.percentile(col, 75)
IQR = Q3 - Q1
print(f"variance: {variance:.2f}")
print(f"range: {data_range:.2f}")
print(f"std_dev: {std_dev:.2f}")
```
